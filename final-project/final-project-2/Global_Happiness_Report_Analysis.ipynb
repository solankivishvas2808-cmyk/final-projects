# Global Happiness Report Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set_theme()

print("Libraries imported successfully!")
```

## 1. Load the Dataset

```python
FILE_NAME = "2015-selected-columns.csv"

df = pd.read_csv(FILE_NAME)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

## 2. Explore the Dataset

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum())
```

## 3. Statistical Summary

```python
display(df.describe())
```

## 4. Top 10 Happiest Countries

```python
top10 = df.sort_values("Happiness Score", ascending=False).head(10)

display(top10[["Country", "Happiness Score"]])

plt.figure(figsize=(12, 6))
sns.barplot(data=top10, x="Happiness Score", y="Country")
plt.title("Top 10 Happiest Countries - 2015")
plt.xlabel("Happiness Score")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## 5. Bottom 10 Countries by Happiness Score

```python
bottom10 = df.sort_values("Happiness Score").head(10)

display(bottom10[["Country", "Happiness Score"]])

plt.figure(figsize=(12, 6))
sns.barplot(data=bottom10, x="Happiness Score", y="Country")
plt.title("Bottom 10 Countries by Happiness Score - 2015")
plt.xlabel("Happiness Score")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## 6. Distribution of Happiness Scores

```python
plt.figure(figsize=(10, 6))
sns.histplot(df["Happiness Score"], bins=20, kde=True)
plt.title("Distribution of Happiness Scores")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.tight_layout()
plt.show()
```

## 7. GDP per Capita vs Happiness

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Economy (GDP per Capita)",
    y="Happiness Score"
)
plt.title("GDP per Capita vs Happiness Score")
plt.xlabel("Economy (GDP per Capita)")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

gdp_corr = df["Economy (GDP per Capita)"].corr(df["Happiness Score"])
print(f"Correlation between GDP and Happiness: {gdp_corr:.3f}")
```

## 8. Social Support (Family) vs Happiness

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Family",
    y="Happiness Score"
)
plt.title("Social Support (Family) vs Happiness Score")
plt.xlabel("Family / Social Support")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

family_corr = df["Family"].corr(df["Happiness Score"])
print(f"Correlation between Family/Social Support and Happiness: {family_corr:.3f}")
```

## 9. Life Expectancy vs Happiness

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Health (Life Expectancy)",
    y="Happiness Score"
)
plt.title("Life Expectancy vs Happiness Score")
plt.xlabel("Health (Life Expectancy)")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

health_corr = df["Health (Life Expectancy)"].corr(df["Happiness Score"])
print(f"Correlation between Life Expectancy and Happiness: {health_corr:.3f}")
```

## 10. Freedom vs Happiness

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Freedom",
    y="Happiness Score"
)
plt.title("Freedom vs Happiness Score")
plt.xlabel("Freedom")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

freedom_corr = df["Freedom"].corr(df["Happiness Score"])
print(f"Correlation between Freedom and Happiness: {freedom_corr:.3f}")
```

## 11. Government Trust vs Happiness

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Trust (Government Corruption)",
    y="Happiness Score"
)
plt.title("Government Trust vs Happiness Score")
plt.xlabel("Trust (Government Corruption)")
plt.ylabel("Happiness Score")
plt.tight_layout()
plt.show()

trust_corr = df["Trust (Government Corruption)"].corr(df["Happiness Score"])
print(f"Correlation between Government Trust and Happiness: {trust_corr:.3f}")
```

## 12. Correlation Heatmap

```python
numeric_columns = [
    "Happiness Rank",
    "Happiness Score",
    "Standard Error",
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom",
    "Trust (Government Corruption)"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap - World Happiness Report 2015")
plt.tight_layout()
plt.show()
```

## 13. Regional Happiness Analysis

```python
regional_happiness = (
    df.groupby("Region")["Happiness Score"]
      .mean()
      .sort_values(ascending=False)
)

print("Average Happiness Score by Region:")
display(regional_happiness.to_frame("Average Happiness Score"))

plt.figure(figsize=(12, 7))
regional_happiness.plot(kind="bar")
plt.title("Average Happiness Score by Region")
plt.xlabel("Region")
plt.ylabel("Average Happiness Score")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## 14. Correlation of Major Factors with Happiness

```python
factors = [
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom",
    "Trust (Government Corruption)"
]

factor_correlations = (
    df[factors + ["Happiness Score"]]
    .corr()["Happiness Score"]
    .drop("Happiness Score")
    .sort_values(ascending=False)
)

display(factor_correlations.to_frame("Correlation with Happiness"))
```

## 15. insight

- Happiness scores vary considerably among countries.
- Economic conditions show a positive association with happiness.
- Social support/family is an important factor associated with happiness.
- Health/life expectancy also shows a positive relationship with happiness.
- Freedom and government trust can be compared with happiness using correlation and visualizations.
- Regional analysis helps identify geographical differences.
- **Important:** Correlation shows association; it does not prove that one factor directly causes happiness.
